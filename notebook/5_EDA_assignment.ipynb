#### ● Detect and treat outliers if necessary.

```python
num_cols = data.select_dtypes(include=np.number).columns
for col in num_cols:
    plt.figure(figsize=(6,3))
    plt.boxplot(data[col])
    plt.title(col)
    plt.show()
```

#### Handle outliers using IQR method

```python
for col in num_cols:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR

    data = data[(data[col] >= lower) & (data[col] <= upper)]
```

```python
num_cols = data.select_dtypes(include=np.number).columns
for col in num_cols:
    plt.figure(figsize=(6,3))
    plt.boxplot(data[col])
    plt.title(col)
    plt.show()
```

- Box plots are used to visually identify outliers and unusual values in the numerical variables. These plots help understand whether some observations are significantly different from the rest of the data.
